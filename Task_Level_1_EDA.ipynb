Import Packages

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Import Dataset

```python
df = pd.read_csv('/content/1) iris.csv')
```

Descriptive Statistic

```python
df.describe()
```

Data Visualization

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df, x='sepal_length', hue='species', kde=True, element="step")
plt.title('Distribution of Sepal Length by Species')
plt.show()
```

```python
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')
plt.title('Sepal Length vs Sepal Width by Species')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species')
plt.title('Petal Length vs Petal Width by Species')
plt.show()
```

```python
sns.pairplot(df, hue='species', markers=["o", "s", "D"])
plt.suptitle('Pairplot Hubungan Antar Fitur', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='species', y='petal_length')
plt.title('Boxplot Petal Length by Species')
plt.show()
```

```python
numeric_df = df.drop(columns=['species'])
correlation_matric = numeric_df.corr()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(correlation_matric, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```
